A large proportion of games do not contain review data.

For analysis involving review-based metrics such as positive review rate, games with zero total reviews are treated separately to avoid misleading calculations.

The positive review rate is calculated as:

Positive Reviews / (Positive Reviews + Negative Reviews) × 100

Games with zero positive and zero negative reviews do not provide enough information to calculate a meaningful review percentage.

The dataset contains over 114,000 Steam games and provides a large amount of information for marketplace and player-engagement analysis.

The dataset has no detected duplicate records, AppID is unique, and the core analytical variables are complete.

Although several metadata fields contain missing values, these fields are not essential for the main analysis.

The presence of outliers, particularly in player engagement and review-related variables, reflects the highly uneven nature of the Steam marketplace and was therefore investigated rather than automatically removed.

Overall, the dataset is suitable for Exploratory Data Analysis and Power BI dashboard development.

# Key Findings

### 1. Steam has a highly concentrated player-engagement distribution

Peak CCU is extremely right-skewed. Most games have relatively low concurrent player activity, while a small number of highly successful games attract exceptionally large player populations.

The maximum recorded Peak CCU in the dataset is approximately 1.01 million players. Because of these extreme values, average Peak CCU should be interpreted alongside the median and, where appropriate, using logarithmic visualizations.

### 2. A small number of games dominate player activity

Counter-Strike 2 recorded the highest Peak CCU in the dataset at approximately 1.01 million concurrent players, followed by Dota 2 at approximately 624 thousand and PUBG: BATTLEGROUNDS at approximately 315 thousand.

This demonstrates the highly concentrated nature of player engagement on Steam, where a relatively small group of major titles attracts a very large share of concurrent players.

### 3. Free games show higher average engagement among active games

After filtering for games with recorded player activity (Peak CCU > 0), free games had an average Peak CCU of approximately 1,500, compared with approximately 190 for paid games.

This suggests that free-to-play games in this dataset tend to achieve substantially higher concurrent player activity than paid games.

However, this represents an association and does not establish that being free directly causes higher engagement.

### 4. Total reviews have a strong positive relationship with player engagement

Among games with both reviews and recorded player activity, Total Reviews and Peak CCU showed a strong positive correlation of approximately 0.879.

This indicates that games with larger numbers of reviews generally tend to have higher concurrent player activity.

The relationship should be interpreted as an association rather than causation, since both variables may be influenced by overall game popularity and exposure.

### 5. Recommendations are also strongly associated with player engagement

After correctly filtering the dataset to games with both Recommendations > 0 and Peak CCU > 0, the correlation between Recommendations and Peak CCU was approximately 0.716.

This indicates a strong positive relationship: games receiving more recommendations generally tend to have higher concurrent player activity.

### 6. Positive review rate has little relationship with Peak CCU

The correlation between Positive Review Rate and Peak CCU was approximately 0.051.

This indicates a very weak linear relationship between the proportion of positive reviews and concurrent player activity.

Therefore, a game having a high positive review percentage does not necessarily mean that it will have a large concurrent player population.

### 7. Player engagement varies substantially across genres

Among games with recorded player activity, Photo Editing, Free To Play, and Massively Multiplayer were among the genres with the highest average Peak CCU.

Photo Editing recorded the highest average Peak CCU at approximately 2,854, followed by Free To Play at approximately 1,937 and Massively Multiplayer at approximately 1,836.

These averages should be interpreted carefully because some genres contain considerably fewer games than others and averages can be strongly affected by highly successful titles.

### 8. Steam's game catalog has expanded significantly over time

The number of games in the dataset increased substantially across recent release years.

For example, the dataset contains approximately 8,723 games released in 2020, compared with approximately 21,636 games released in 2025.

This indicates substantial growth in the number of games being added to the Steam marketplace.

### 9. Release-year engagement is highly uneven

Among games with recorded player activity, average Peak CCU varies considerably across release years.

The 2012 release cohort recorded an unusually high average Peak CCU of approximately 5,303, while its median Peak CCU was only 5.

This large difference between the mean and median indicates that the 2012 average is heavily influenced by a relatively small number of highly successful games.

### 10. Recent release cohorts generally contain many games with relatively low typical engagement

From 2014 onward, the number of games in each release cohort increased substantially, while median Peak CCU remained relatively low.

This suggests that although the Steam catalog has expanded rapidly, most individual games do not achieve the very high levels of player activity reached by the platform's most successful titles.

Overall, the analysis indicates that Steam's player engagement is highly concentrated, with a relatively small number of games accounting for exceptionally high concurrent player activity.

# Conclusion

This analysis explored the Steam gaming marketplace using a dataset containing more than 114,000 games. The analysis examined pricing, reviews, recommendations, player engagement, genres, free-to-play status, and release-year trends to understand the factors associated with game popularity and player activity.

The analysis found that player engagement on Steam is highly uneven. Most games have relatively low Peak CCU, while a small number of highly successful games attract exceptionally large concurrent player populations. This concentration is clearly visible in the distribution of Peak CCU and in the ranking of the most popular games.

Free games showed substantially higher average Peak CCU than paid games among games with recorded player activity. In addition, Total Reviews and Recommendations both showed strong positive relationships with Peak CCU, suggesting that popular games tend to accumulate greater levels of community engagement.

However, Positive Review Rate showed almost no linear relationship with Peak CCU. This indicates that having a high proportion of positive reviews alone does not necessarily correspond to high concurrent player activity.

Genre-level analysis also showed considerable differences in average engagement. Certain smaller genres recorded very high average Peak CCU, while more common genres such as Action and RPG had lower averages. These results demonstrate the importance of considering sample size and the influence of outliers when comparing genres.

The release-year analysis showed substantial growth in the Steam game catalog over time. At the same time, typical player engagement remained relatively low for most games, highlighting the highly competitive nature of the Steam marketplace.

Overall, the analysis demonstrates that Steam's marketplace is characterized by rapid catalog growth and highly concentrated player engagement. A small number of successful games achieve exceptionally high player activity, while the majority of games operate at much lower engagement levels.

The findings from this exploratory analysis provide a foundation for further business-oriented analysis and visualization through an interactive dashboard.
